### Q15: Ek variable count = 0 banayen. List ['apple', 'mango', 'apple', 'banana', 'apple'] par loop lagayen aur count karein ke 'apple' kitni baar aya hai.

### Q20: Password check: List ['123', 'abc', 'admin123', 'xyz'] par loop lagayen. Agar 'admin123' mil jaye tou 'Access Granted' print kar ke loop break kar dein.

```python
password=['123', 'abc', 'admin123', 'xyz']
for x in password:
    if x =='admin123':
        print('Access Granted')
        break
    else:
        print('Access Denied')
```
